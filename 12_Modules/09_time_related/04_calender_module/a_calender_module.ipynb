Problem: For a given date, tell the week of the day

```python
import calendar


def get_week_of_day():
    given_input = input('Enter the date in "MM DD YYYY" format:')
    month, day, year = map(int, given_input.split(" "))

    weeks = (
        "MONDAY",
        "TUESDAY",
        "WEDNESDAY",
        "THURSDAY",
        "FRIDAY",
        "SATURDAY",
        "SUNDAY",
    )
    week_no = calendar.weekday(year, month, day)
    week_name = weeks[week_no]
    return f"day of week for {given_input} is {week_name}"
```

```python
get_week_of_day()
```

itermonthdates() - returns an iterator of all days of the given month. - Also, all days before the start of the month and after the end of the month, required to get the complete week, are displayed.

```python
c = calendar.Calendar()
for i in c.itermonthdates(2020, 1):
    print(i, end="\t")
```

itermonthdays() - similar to itermonthdates method, but it only returns the day numbers.

```python
c = calendar.Calendar()
for i in c.itermonthdays(2020, 1):
    print(i, end=" ")
```

itermonthdays2() - displays a tuple consisting of day and weekday numbers.

```python
c = calendar.Calendar()
for i in c.itermonthdays2(2020, 1):
    print(i, end="\t")
```

itermonthdays3() - similar to the itermonthdays3() method, except that it returns a tuple of year, month, and the day of the month.

```python
c = calendar.Calendar()
for i in c.itermonthdays3(2020, 1):
    print(i, end="\t")
```

monthdatescalendar() - returns a list of full weeks in the month - Each week is a list of 7 datetime.date objects.

```python
c = calendar.Calendar()
for i in c.monthdatescalendar(2020, 1):
    print(i)
```

monthdays2calendar() - returns a list of weeks, with each week as 7 tuples of the day of month and day of the week. - the first value of the tuple is the day of the month (0-31) and second value of the tuple is the week number (0-6)

```python
c = calendar.Calendar()
for i in c.monthdays2calendar(2020, 1):
    print(i)
```

onthdayscalendar() - returns a list of full weeks, with each week being a list of days of a month.

```python
c = calendar.Calendar()
for i in c.monthdayscalendar(2020, 1):
    print(i)
```

```python
c = calendar.Calendar()
for i in c.yeardatescalendar(2020, 1):
    for month in i:
        for week in month:
            for day in week:
                print(day, end="\t")
    print("\n\n")
```
